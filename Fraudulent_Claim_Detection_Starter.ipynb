## **Fraudulent Claim Detection**

## Problem Statement
Global Insure, a leading insurance company, processes thousands of claims annually. However, a significant percentage of these claims turn out to be fraudulent, resulting in considerable financial losses. The company’s current process for identifying fraudulent claims involves manual inspections, which is time-consuming and inefficient. Fraudulent claims are often detected too late in the process, after the company has already paid out significant amounts. Global Insure wants to improve its fraud detection process using data-driven insights to classify claims as fraudulent or legitimate early in the approval process. This would minimise financial losses and optimise the overall claims handling process.

## Business Objective
Global Insure wants to build a model to classify insurance claims as either fraudulent or legitimate based on historical claim details and customer profiles. By using features like claim amounts, customer profiles and claim types, the company aims to predict which claims are likely to be fraudulent before they are approved.

## **2. Data Cleaning** <font color = red>[10 marks]</font>

### **2.1 Handle null values** <font color = red>[2 marks]</font>

#### **2.1.1** Examine the columns to determine if any value or column needs to be treated <font color="red">[1 Mark]</font>

```python
# Check the number of missing values in each column

# Replace '?' with None
df['collision_type'] = df['collision_type'].map(lambda x: None if x == '?' else x)
df['property_damage'] = df['property_damage'].map(lambda x: None if x == '?' else x)
df['police_report_available'] = df['police_report_available'].map(lambda x: None if x == '?' else x)

null_counts = df.isnull().sum()
null_columns = null_counts[null_counts > 0]
print(null_columns)
```

```python
# Rows where authorities_contacted is null
mask = df['authorities_contacted'].isna()
subset = df.loc[mask]

print(f"Rows with authorities_contacted null: {len(subset)}")
print("\npolice_report_available for those rows:")
print(subset['police_report_available'].value_counts(dropna=False))
```

```python
# find the mode of null_columns (use .index to get column names; null_columns is a Series of counts)
mode_df = df[null_columns.index].mode()
mode_values = mode_df.iloc[0] if len(mode_df) > 0 else pd.Series(pd.NA, index=null_columns.index)

mode_values
```

#### **2.1.2** Handle rows containing null values <font color="red">[1 Mark]</font>

```python
# Handle the rows containing null values

# Assuming if police_report_available then authorities_contacted with null value should have value 'Police'
df.loc[df['police_report_available'].isna(), 'authorities_contacted'] = 'Police'

# For rows with null value in column: 'property_damage', if 'property_claim' is 0 then set 'property_damage' to 'NO', else set it to 'YES'
df.loc[df['property_damage'].isna() & (df['property_claim'] == 0), 'property_damage'] = 'NO'
df.loc[df['property_damage'].isna() & (df['property_claim'] > 0), 'property_damage'] = 'YES'

# for null values of collision_type,  set it to Rear Collision (which is also mode for collision_type)
df['collision_type'].fillna('Rear Collision', inplace=True)

# for null values of authorities_contacted, populate Other
df['authorities_contacted'].fillna('Other', inplace=True)

# for remaining null values of police_report_available, populate NO
df['police_report_available'].fillna('NO', inplace=True)
```

```python
null_counts = df.isnull().sum()
null_columns = null_counts[null_counts > 0]
print(null_columns)
```

### **2.2 Identify and handle redundant values and columns** <font color = red>[5 marks]</font>

#### **2.2.1** Examine the columns to determine if any value or column needs to be treated <font color="red">[2 Mark]</font>

```python
# Write code to display all the columns with their unique values and counts and check for redundant values

# create a data frame with column names and their unique values, redundant values and their counts
unique_values = []
for column in df.columns:
    unique_values.append({
        'column': column,
        'unique_count': df[column].nunique(),
        'redundant_values': df[column].value_counts().index[df[column].value_counts() > 1].tolist(),
        'redundant_value_range': (df[column].min(), df[column].max()),
        'redundant_counts': df[column].value_counts()[df[column].value_counts() > 1].tolist()
    })
unique_values_df = pd.DataFrame(unique_values).sort_values(by='unique_count', ascending=False)
unique_values_df
```

```python
# Identify and drop any columns that are completely empty

# drop _c39 column as it has all null values
df = df.drop(columns=['_c39'])
```

#### **2.2.3** Identify and drop rows where features have illogical or invalid values, such as negative values for features that should only have positive values <font color="red">[1 Mark]</font>

```python
# Identify and drop rows where features have illogical or invalid values, such as negative values for features that should only have positive values

#drop row where umbrella_limit is negative
df = df[df['umbrella_limit'] >= 0]

# Some data was find where claim was 100 and annual premium was more than claim amount which does not make sense, droppig such rows
df = df[df['policy_annual_premium'] < df['total_claim_amount']]
```

#### **2.2.4** Identify and remove columns where a large proportion of the values are unique or near-unique, as these columns are likely to be identifiers or have very limited predictive power <font color="red">[1 Mark]</font>

```python
# Identify and remove columns that are likely to be identifiers or have very limited predictive power

# drop policy_number as it is an identifier and has no predictive power
df = df.drop(columns=['policy_number'])

# drop incident_location as all rows have unique values and it is unlikely to have predictive power
df = df.drop(columns=['incident_location'])

# drop insured_zip as 99 percent rows have unique values and it is unlikely to have predictive power
df = df.drop(columns=['insured_zip'])
```

#### **4.4.1** Target likelihood analysis for categorical variables. <font color="red">[5 Marks]</font>
Investigate the relationships between categorical features and the target variable by analysing the target event likelihood (for the `'Y'` event) for each level of every relevant categorical feature. Through this analysis, identify categorical features that do not contribute much in explaining the variation in the target variable.

```python
# Write a function to calculate and analyse the target variable likelihood for categorical features
def plot_categorical_target_likelihood(data, target, categorical_cols):
    num_cols = len(categorical_cols)
    num_rows = (num_cols + 2) // 3  # Calculate number of rows needed for 3 columns per row
    plt.figure(figsize=(15, num_rows * 5))  # Adjust height based on number of rows

    for i, col in enumerate(categorical_cols):
        plt.subplot(num_rows, 3, i + 1)  # Create subplot
        # Calculate target variable likelihood for each category
        likelihood = data.groupby(col)[target].mean()
        likelihood.plot(kind='bar')  # Plot bar chart
        plt.title(f'Target Variable Likelihood by {col}')  # Set title
        plt.xlabel(col)  # Set x-axis label
        plt.ylabel('Likelihood of Fraud Reported')  # Set y-axis label

    plt.tight_layout()  # Adjust layout to prevent overlap
    plt.show()



```

```python
# Select categorical columns
categorical_cols = X_train.select_dtypes(include=['object']).columns.tolist()

# Add numeric target as a new column to X_train
X_train_with_target = X_train.copy()
X_train_with_target['fraud_reported_numeric'] = y_train.map({'Y': 1, 'N': 0})

# Plot target variable likelihood for categorical features
plot_categorical_target_likelihood(X_train_with_target, 'fraud_reported_numeric', categorical_cols)
```

### **6.1 Perform resampling** <font color="red">[3 Marks]</font>

Handle class imbalance in the training data by applying resampling technique.

Use the **RandomOverSampler** technique to balance the data and handle class imbalance. This method increases the number of samples in the minority class by randomly duplicating them, creating synthetic data points with similar characteristics. This helps prevent the model from being biased toward the majority class and improves its ability to predict the minority class more accurately.

**Note:** You can try other resampling techniques to handle class imbalance

```python
# Import RandomOverSampler from imblearn library
from imblearn.over_sampling import RandomOverSampler

# Perform resampling on training data
ros = RandomOverSampler(random_state=42)
X_train_resampled, y_train_resampled = ros.fit_resample(X_train, y_train)

```

### **6.2 Feature Creation** <font color="red">[4 marks]</font>

Create new features from existing ones to enhance the model's ability to capture patterns in the data. This may involve deriving features from date/time columns, combining features, or creating interaction terms.

```python
X_train_resampled['insured_occupation'].value_counts()
```

```python
# Create new features based on your understanding for both training and validation data


#Extract Policy bind month as it may have an impact on the target variable as certain months may have more fraudulent claims than others
X_train_resampled['policy_bind_month'] = X_train_resampled['policy_bind_date'].dt.month
X_val['policy_bind_month'] = X_val['policy_bind_date'].dt.month


#Create a feature to provide number of days between policy bind date and incident date as it may have an impact on the target variable as people with fraudulent intent may file claim immediately after buying insurance policy
X_train_resampled['days_between_policy_bind_and_incident'] = (X_train_resampled['incident_date'] - X_train_resampled['policy_bind_date']).dt.days
X_val['days_between_policy_bind_and_incident'] = (X_val['incident_date'] - X_val['policy_bind_date']).dt.days


#Create incident day of week
X_train_resampled['incident_day_of_week'] = X_train_resampled['incident_date'].dt.dayofweek
X_val['incident_day_of_week'] = X_val['incident_date'].dt.dayofweek

#claim to premium ratio
X_train_resampled['claim_to_premium_ratio'] = X_train_resampled['total_claim_amount'] / X_train_resampled['policy_annual_premium']
X_val['claim_to_premium_ratio'] = X_val['total_claim_amount'] / X_val['policy_annual_premium']


#Time of day bucket (0-5), (6-11), (12-17), (18-23)
X_train_resampled['incident_hour_bucket'] = pd.cut(X_train_resampled['incident_hour_of_the_day'], bins=[-1, 5, 11, 17, 23], labels=['0-5', '6-11', '12-17', '18-23'])
X_val['incident_hour_bucket'] = pd.cut(X_val['incident_hour_of_the_day'], bins=[-1, 5, 11, 17, 23], labels=['0-5', '6-11', '12-17', '18-23'])


#Create a boolean feature indicating either the user is highly educated (Masters, PhD or MD) or not as it may have an impact on the target variable
X_train_resampled['highly_educated'] = X_train_resampled['insured_education_level'].isin(['Masters', 'PhD', 'MD'])
X_val['highly_educated'] = X_val['insured_education_level'].isin(['Masters', 'PhD', 'MD'])



#Create Vehicle age as difference between incident year and auto_year and drop auto_year
X_train_resampled['vehicle_age'] = X_train_resampled['incident_date'].dt.year - X_train_resampled['auto_year']
X_val['vehicle_age'] = X_val['incident_date'].dt.year - X_val['auto_year']



#Create net capital gain as sum of capital-gains and capital-loss and drop capital-gains and capital-loss
X_train_resampled['net_capital_gain'] = X_train_resampled['capital-gains'] + X_train_resampled['capital-loss']
X_val['net_capital_gain'] = X_val['capital-gains'] + X_val['capital-loss']
X_train_resampled = X_train_resampled.drop(columns=['capital-gains', 'capital-loss'])
X_val = X_val.drop(columns=['capital-gains', 'capital-loss'])


```

```python

# ─────────────────────────────────────────────────────────────────────────────
# ADDITIONAL INTERACTION & DOMAIN-DRIVEN FEATURES
# Each feature is grounded in EDA findings from Sections 4 and 5
# ─────────────────────────────────────────────────────────────────────────────

for df_feat in [X_train_resampled, X_val]:

    # 1. HIGH-RISK HOBBY FLAG
    # EDA finding: ~80% of fraud cases had chess or cross-fit as hobby.
    # This single binary flag captures a very strong fraud signal.
    df_feat['is_high_risk_hobby'] = df_feat['insured_hobbies'].isin(['chess', 'cross-fit']).astype(int)

    # 2. HIGH-RISK OCCUPATION FLAG
    # EDA finding: transport-moving and exec-managerial showed elevated fraud likelihood.
    df_feat['is_high_risk_occupation'] = df_feat['insured_occupation'].isin(
        ['exec-managerial']
    ).astype(int)


    # 7. VEHICLE-TO-INJURY CLAIM RATIO
    # EDA finding: vehicle_claim is the biggest driver of total claim in fraudulent cases.
    # A disproportionately high vehicle claim relative to injury claim signals inflated vehicle damage.
    df_feat['vehicle_to_injury_ratio'] = df_feat['vehicle_claim'] / (df_feat['injury_claim'] + 1)

    # 8. EARLY CLAIM FLAG
    # Fraudsters often file claims soon after purchasing a policy.
    # Flag incidents occurring within the first year of policy binding.
    df_feat['early_claim_flag'] = (df_feat['days_between_policy_bind_and_incident'] < 365).astype(int)

    # 9. HIGH-RISK STATE FLAG
    # EDA finding: incident states OH and PA had noticeably higher fraud likelihood.
    df_feat['is_high_risk_state'] = df_feat['incident_state'].isin(['OH', 'PA']).astype(int)

    # 10. HIGH-RISK AUTO MAKE FLAG
    # EDA finding: Audi, Ford, Mercedes showed higher fraud likelihood.
    df_feat['is_high_risk_auto_make'] = df_feat['auto_make'].isin([ 'Ford', 'Mercedes']).astype(int)

print("New interaction features added successfully.")
print("New columns:", [
    'is_high_risk_hobby', 'is_high_risk_occupation', 'educated_high_risk_occ',
    'damage_no_police_report', 'is_major_damage', 'vehicle_claim_per_vehicle',
    'vehicle_to_injury_ratio', 'early_claim_flag', 'is_high_risk_state', 'is_high_risk_auto_make'
])
```

### **7.3 Find the Optimal Cutoff** <font color = red>[12 marks]</font>
Find the optimal cutoff to improve model performance by evaluating various cutoff values and their impact on relevant metrics.

#### **7.3.1** Plot ROC Curve  to visualise the trade-off between true positive rate and false positive rate across different classification thresholds <font color="red">[2 Marks]</font>

```python
# Import libraries or function to plot the ROC curve

import matplotlib.pyplot as plt

# Define ROC function
def plot_roc_curve(y_true, y_pred_prob):
    fpr, tpr, thresholds = metrics.roc_curve(y_true, y_pred_prob)
    auc_roc = metrics.roc_auc_score(y_true, y_pred_prob)

    plt.figure(figsize=(8, 6))
    plt.plot(fpr, tpr, label=f'ROC Curve (AUC = {auc_roc:.4f})')
    plt.plot([0, 1], [0, 1], 'k--')  # Diagonal line for random classifier
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Receiver Operating Characteristic (ROC) Curve')
    plt.legend(loc='lower right')
    plt.grid()
    plt.show()
```

**Sensitivity and Specificity tradeoff**

After analysing the area under the curve of the ROC, check the sensitivity and specificity tradeoff to find the optimal cutoff point.

#### **7.3.2** Predict on training data at various probability cutoffs <font color="red">[1 Mark]</font>

```python
# Create columns with different probability cutoffs to explore the impact of cutoff on model performance
cutoffs = [0.2, 0.3, 0.38, 0.4, 0.45, 0.475, 0.5, 0.6, 0.7, 0.8, 0.9]
```

#### **7.3.3** Plot accuracy, sensitivity, specificity at different values of probability cutoffs <font color="red">[2 Marks]</font>

```python
# Create a DataFrame to see the values of accuracy, sensitivity, and specificity at different values of probability cutoffs
rows = []

for cutoff in cutoffs:
    train_results['Predicted'] = (train_results['Predicted_Prob'] >= cutoff).astype(int)
    conf_matrix = metrics.confusion_matrix(train_results['Actual'], train_results['Predicted'])
    tn, fp, fn, tp = conf_matrix.ravel()

    accuracy = (tp + tn) / (tp + tn + fp + fn)
    sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = sensitivity
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

    rows.append({'Cutoff': cutoff, 'Accuracy': accuracy, 'Sensitivity': sensitivity,
                 'Specificity': specificity, 'Precision': precision, 'Recall': recall, 'F1 Score': f1})

cutoff_analysis = pd.DataFrame(rows)
cutoff_analysis
```

## 4. Business Conclusions & Recommendations

### 💡 Insight 1: Fraud is Strongly Linked to Hobby Risk Profiles
The `is_high_risk_hobby` feature was the **single strongest predictor** (coef = +2.23, highly significant). Insurance underwriters should incorporate lifestyle and hobby data into risk scoring at the policy issuance stage. Policyholders with extreme or high-risk hobbies should undergo enhanced claim scrutiny.

### 💡 Insight 2: Incident Severity is a Critical Discriminator
Claims with **Major Damage** were significantly more likely to be fraudulent compared to Minor, Trivial, or Total Loss incidents. This counter-intuitive finding suggests that fraudsters tend to fabricate moderate-severity incidents (Major Damage) to remain below the radar of large-payout investigations, while genuine incidents tend to be at the extremes (minor fender-benders or genuine total losses).

### 💡 Insight 3: Vehicle-to-Injury Ratio Flags Staged Accidents
A high ratio of vehicle claim to injury claim is a classic indicator of a staged accident where vehicles are deliberately damaged but minimal physical injury occurs. Automated alert rules can be implemented in the claims management system for claims where `vehicle_claim > 3× injury_claim`.

### 💡 Insight 4: Relationship to Policyholder Matters
Claims filed by **other-relatives** and **non-family members** carry a higher fraud risk. This may indicate organised fraud rings using a single policy to make multiple claims through a network of associates.

### 💡 Insight 5: Witnesses and Property Damage Together Raise Flags
The combination of multiple witnesses with reported property damage is associated with staged collisions. Investigators should be alerted when both `witnesses ≥ 2` and `property_damage = YES` are present without a corresponding police report.

### 💡 Insight 6: Geographic Concentration Exists
Incident states **VA (Virginia)** showed elevated fraud probability, while **WV (West Virginia)** showed lower risk. Deploying additional investigation resources in high-risk regions can improve detection efficiency.

### 💡 Insight 7: Early Claims Flag Potential Policy Fraud
The `early_claim_flag` — where a claim is filed within a very short period after policy binding — is a well-known fraud indicator. Implementing a mandatory waiting period or enhanced review for early claims is recommended.
